```python

import numpy as np
import matplotlib.pyplot as plt


k_B = 1.380649e-23       # Constante de Boltzmann (J/K)
m_He = 6.646476e-27      # Masa de un átomo de helio (kg)
T = 400                  # Temperatura (K)


def maxwell(v, m, T):
    factor = 4 * np.pi * (m / (2 * np.pi * k_B * T))**(3/2)
    exponential = np.exp(-m * v**2 / (2 * k_B * T))
    return factor * v**2 * exponential


v_probable = np.sqrt(2 * k_B * T / m_He)

v_media = np.sqrt(8 * k_B * T / (np.pi * m_He))

v_rms = np.sqrt(3 * k_B * T / m_He)


v = np.linspace(0, 3000, 1000)
f = maxwell(v, m_He, T)


plt.figure(figsize=(10, 6))

plt.plot(v, f, label='Helio a 400 K', linewidth=2)

plt.axvline(v_probable, linestyle='--',
            label='Velocidad más probable')

plt.axvline(v_media, linestyle='--',
            label='Velocidad media')

plt.axvline(v_rms, linestyle='--',
            label='Velocidad RMS')

plt.xlabel('Rapidez (m/s)')
plt.ylabel('Densidad de probabilidad (s/m)')
plt.title('Distribución de velocidades de Maxwell del helio (400 K)')

plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("DISTRIBUCIÓN DE MAXWELL PARA EL HELIO")
print(f"Temperatura: {T} K")
print("--------------------------------------")
print(f"Velocidad más probable: {v_probable:.2f} m/s")
print(f"Velocidad media:        {v_media:.2f} m/s")
print(f"Velocidad RMS:          {v_rms:.2f} m/s")

# ==========================================
# 7. TABLA DE DATOS
# ==========================================

print("\nTABLA DE DISTRIBUCIÓN")
print("Velocidad (m/s) | f(v) (s/m)")

for velocidad in range(0, 3001, 250):
    probabilidad = maxwell(velocidad, m_He, T)
    print(f"{velocidad:15d} | {probabilidad:.6e}")
```
